```python
# Data Loading and Preparation
import pandas as pd
import numpy as np
import tensorflow as tf

## Load MNIST dataset using Keras
(X_train_np, y_train_np), (X_test_np, y_test_np) = tf.keras.datasets.mnist.load_data()

# Concatenate training and test data
X_full_np = np.concatenate((X_train_np, X_test_np), axis=0)
y_full_np = np.concatenate((y_train_np, y_test_np), axis=0)

# Reshape X from (N, 28, 28) to (N, 784) and convert to DataFrame
X = pd.DataFrame(X_full_np.reshape(X_full_np.shape[0], -1))

# Convert y to pandas Series of strings to match fetch_openml output type
y = pd.Series(y_full_np.astype(str))
```

```python
# Normalize the data
X = X / 255.0
```

```python
# Check Data Shape
print("Dataset shape:", X.shape)
```

```python
# Exploratory Data Analysis (EDA)
## Visualize sample image
import matplotlib.pyplot as plt

plt.imshow(X.iloc[0].values.reshape(28, 28))
plt.title(f"Label: {y[0]}")
plt.show()
```

```python
# Clustering with K-Means
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=10, random_state=42)
kmeans.fit(X)
```

```python
# Cluster Labels
clusters = kmeans.labels_
```

```python
# Find Optimal Clusters (Elbow Method)
inertia = []

for k in range(1, 15):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertia.append(km.inertia_)

# Plot
plt.plot(range(1, 15), inertia, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.show()
```

```python
# Analyze Cluster Centroids
centroids = kmeans.cluster_centers_

plt.imshow(centroids[0].reshape(28, 28))
plt.title("Cluster Center")
plt.show()
```

```python
# Predict new data
new_image = X.iloc[1].values.reshape(1, -1)
predicted_cluster = kmeans.predict(new_image)
print("Predicted Cluster:", predicted_cluster)
```

```python
# Dimensionality Reduction
## PCA (Linear Reduction)
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
```

```python
# Dimensionality Reduction
## t-SNE (non-linear)
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X.sample(5000))
```

```python
# Visualization
## Visualize PCA
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, s=1)
plt.title("PCA Clusters")
plt.show()

## Visualize t-SNE
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=clusters[:5000], s=1)
plt.title("t-SNE Clusters")
plt.show()
```
